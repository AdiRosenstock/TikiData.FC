```python
import soccerdata as sd
import pandas as pd

fbref = sd.FBref(
    leagues="Big 5 European Leagues Combined",
    seasons="2025-2026"
)

# Pull each stat type
standard = fbref.read_player_season_stats(stat_type="standard").reset_index()
shooting = fbref.read_player_season_stats(stat_type="shooting").reset_index()
misc = fbref.read_player_season_stats(stat_type="misc").reset_index()

# Merge them
df = standard.merge(
    shooting,
    on=["player", "team", "league", "season"],
    how="left",
    suffixes=("", "_shooting")
)

df = df.merge(
    misc,
    on=["player", "team", "league", "season"],
    how="left",
    suffixes=("", "_misc")
)
```

```python
new_df = df.copy()

new_df.columns = [
    col[0] if col[1] == "" else f"{col[0]}_{col[1]}"
    for col in new_df.columns
]

new_df.columns.tolist()
```

```python
new_df = new_df[
    new_df["pos"] != "GK"
]

new_df["Playing Time_Min"] = pd.to_numeric(new_df["Playing Time_Min"], errors="coerce")
new_df["Performance_Gls"] = pd.to_numeric(new_df["Performance_Gls"], errors="coerce")
new_df["age"] = new_df["age"].astype(str).str.split("-").str[0]
new_df["age"] = pd.to_numeric(new_df["age"], errors="coerce")
```

```python
new_df = new_df.drop(
    columns=[
        # nation duplicates
        "nation_shooting",
        "nation_misc",

        # 90s duplicates
        "90s",
        "90s_misc",

        # goals duplicate
        "Standard_Gls",

        # card duplicates
        "Performance_misc_CrdY",
        "Performance_misc_CrdR",

        "age_shooting",
        "born_shooting",
        "pos_shooting",
        "pos_misc",
        "age_misc",
        "born_misc"
    ],
    errors="ignore"
)
```
